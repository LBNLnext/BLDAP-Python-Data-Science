# Important Vocabulary
 <a name='Section1'></a>

**Null hypothesis:** Your null hypothesis is what you expect to see if the effect you are studying does not exist. For example, if we hypothesize that plants live longer in humid climates, then our null hypothesis would be plants do not live longer in humid climates.

**Alternative hypothesis:** Your alternative hypothesis is what you expect to see if the effect you are studying does exist between your two groups or variables. In the example we gave above, our alternative hypothesis would be that plants live longer in humid climates since that is the effect we are interested in looking into.


**What is p-value?** Essentially, p-value is the probability of seeing the observed data assuming your null hypothesis is true. If the probability you get of the observed data is **less** than 0.05, you can safely reject the null hypothesis and say the alternative hypothesis is statistically significant. Note that 0.05 is a common threshold and it can vary depending on your field or what you are analyzing; however, it is the standard for what we will be using. The lower your p-value is, the more confident you can be about your alternative hypothesis. **P-values are critical because they allow you to draw conclusions based on your data.**
